## GDP by Country

```python
#Read in country-level GDP
Country_GDP = pd.read_csv("gdp-per-capita-worldbank.csv")
Country_GDP.rename(columns = {'Entity': 'country'}, inplace = True)
Country_GDP.rename(columns = {'GDP per capita, PPP (constant 2021 international $)': 'country_gdp'}, inplace = True)
Country_GDP_2021 = Country_GDP[Country_GDP['Year'] == 2021]
Country_GDP_2021=Country_GDP_2021.drop(["Year", "Code"], axis = 1)
Country_GDP_2021.head()
```

```python
#Fix the country names on the GDP file to match the PISA file
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Azerbaijan', 'Baku (Azerbaijan)')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Brunei', 'Brunei Darussalam')

# 'Chinese Taipei' / Taiwan not on the GDP file

Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Czechia', 'Czech Republic')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Hong Kong', 'Hong Kong (China)')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('South Korea', 'Korea')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Macao', 'Macao (China)')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Palestine', 'Palestinian  Authority')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Slovakia', 'Slovak Republic')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Turkey', 'Türkiye')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Ukraine', 'Ukrainian regions (18 of 27)')
Country_GDP_2021['country'] = Country_GDP_2021['country'].replace('Vietnam', 'Viet Nam')
```

```python
Country_GDP_2021.describe()
```

## GDP Spending on Education

```python
#Read in country-level percent of GDP spending on education. Take the most recent data from 2019-2022
Country_GDP_Ed = pd.read_csv("UNESCO_Percent_GDP_Education.csv")
Country_GDP_Ed.rename(columns = {'Country Name': 'country'}, inplace = True)
Country_GDP_Ed = Country_GDP_Ed.drop(Country_GDP_Ed.columns[4:63],axis = 1)
Country_GDP_Ed.head()
```

```python
#coalesce values in specific column order
Country_GDP_Ed['ed_percent_gdp'] = Country_GDP_Ed[['2023', '2022', '2021','2020','2019']].bfill(axis=1).iloc[:, 0]
Country_GDP_Ed.head()
```

```python
Country_GDP_Ed = Country_GDP_Ed.drop(['2023', '2022', '2021','2020','2019', 'Country Code', 'Indicator Name', 'Indicator Code'],axis = 1)
```

```python
#Fix the country names on the GDP file to match the PISA file
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Azerbaijan', 'Baku (Azerbaijan)')

# 'Chinese Taipei' / Taiwan not on the GDP ED file
          
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Czechia', 'Czech Republic')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Hong Kong SAR, China', 'Hong Kong (China)')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Korea, Rep.', 'Korea')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Macao SAR, China', 'Macao (China)')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('West Bank and Gaza', 'Palestinian  Authority')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Turkiye', 'Türkiye')
Country_GDP_Ed['country'] = Country_GDP_Ed['country'].replace('Ukraine', 'Ukrainian regions (18 of 27)')

#The following countries are missing the percentage of GDP spent on education: Brunei Darussalam, 
#Colombia, Kosovo, Montenegro, and North Macedonia

Country_GDP_Ed.head()
```

## Merge Data Files

```python
# Merge files with country name in commong including Math, Reading, Science, Math & Sex, SES, GDP, %GDP for ED
dfCOG = pd.merge(PISA_Country_Codes, PISA_Country_Math, how = 'left', on = 'country')
dfCOG = pd.merge(dfCOG, PISA_Country_Read, how = 'left',  on = 'country')
dfCOG = pd.merge(dfCOG, PISA_Country_Scie, how = 'left',  on = 'country')
dfCOG = pd.merge(dfCOG, PISA_Country_Math_Sex, how = 'left',  on = 'country')
dfCOG = pd.merge(dfCOG, PISA_Country_SES, how = 'left',  on = 'country')
dfCOG = pd.merge(dfCOG, Country_GDP_2021, how = 'left',  on = 'country')
dfCOG = pd.merge(dfCOG, Country_GDP_Ed, how = 'left',  on = 'country')
dfCOG.head()
```

```python
#Merge the proportions which were rolled up from the student data
dfCOG = pd.merge(dfCOG, PISA_Country_Proportions, how = 'left',  on = 'CNTRYID')
dfCOG.head()
```

## Add Constructed Variables for the Analysis

```python
def get_region(country):
    # Western Europe countries
    western_europe = [
        'Belgium', 'Denmark', 'France', 'Germany', 'Iceland', 'Ireland', 'Italy', 'Kosovo',
        'Latvia', 'Lithuania', 'Luxembourg', 'Malta', 'Monaco', 'Netherlands', 'Norway',
        'Poland', 'Portugal', 'Romania', 'Spain', 'Sweden', 'Switzerland', 'United Kingdom', 'Morocco','Finland'
    ]

    # Eastern Europe countries
    eastern_europe = [
        'Albania', 'Armenia', 'Austria', 'Bulgaria', 'Croatia', 'Czech Republic', 'Estonia',
        'Georgia', 'Hungary', 'Kazakhstan', 'Moldova', 'Montenegro', 'North Macedonia', 'Poland',
        'Serbia', 'Slovak Republic', 'Slovenia', 'Ukrainian regions (18 of 27)','Greece'
    ]

    # North America countries
    north_america = ['Canada', 'United States', 'Mexico', 'Costa Rica', 'Panama',
                     'Dominican Republic', 'Guatemala', 'El Salvador','Jamaica']

    # South America countries
    south_america = ['Argentina', 'Chile', 'Colombia', 'Paraguay', 'Peru', 'Brazil', 'Uruguay']

    # Asia countries
    asia = ['Cambodia', 'Chinese Taipei', 'Hong Kong (China)', 'Indonesia', 'Israel', 'Japan',
            'Kazakhstan', 'Korea', 'Malaysia', 'Macao (China)', 'Mongolia', 'Philippines', 'Qatar',
            'Saudi Arabia', 'Singapore', 'Thailand', 'Viet Nam', 'Uzbekistan', 'Brunei Darussalam','Australia', 'New Zealand']

    # Middle East countries
    middle_east = ['Jordan', 'United Arab Emirates', 'Palestinian  Authority','Türkiye','Baku (Azerbaijan)','Cyprus']

    # Check which region the country belongs to
    if country in western_europe:
        return 'Western Europe'
    elif country in eastern_europe:
        return 'Eastern Europe'
    elif country in north_america:
        return 'North America'
    elif country in south_america:
        return 'South America'
    elif country in asia:
        return 'Asia'
    elif country in middle_east:
        return 'Middle East'
    else:
        return 'Unknown'

# Add the 'Region' column to the DataFrame
dfCOG['Geo_Region'] = dfCOG['country'].apply(get_region)
```

```python
#Add constructed variables for minimum proficiency
#Average test score for the country below the minimal standards for proficiency
#Math = <420.07, greater than or equal to 420.07 
math_bins = [0, 420.07, 600]
#Reading = <407.47, greater than or equal to 407.47 
read_bins = [0, 407.47, 600]
#Science = <409.54, greater than or equal to 409.54
scie_bins = [0, 409.54, 600]
#Variable labels
proficiency_labels = ['Below minimum proficiency', 'Met minimum proficiency']
#Create variables 
dfCOG['math_min_proficiency'] = pd.cut(dfCOG['math_all_mean'], bins=math_bins, labels=proficiency_labels, right=True)
dfCOG['read_min_proficiency'] = pd.cut(dfCOG['read_all_mean'], bins=read_bins, labels=proficiency_labels, right=True)
dfCOG['scie_min_proficiency'] = pd.cut(dfCOG['scie_all_mean'], bins=scie_bins, labels=proficiency_labels, right=True)
```
